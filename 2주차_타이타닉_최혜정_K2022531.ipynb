```python
import matplotlib.pyplot as plt
 
#시각화 준비
plt.style.use('seaborn')
sns.set(font_scale=1)
sns.set_style("ticks")
sns.set_palette("Paired")

# 한글 사용 준비
plt.rc('font', family='NanumBarunGothic') 

fig, ax = plt.subplots(1, 2, figsize=(14,5))
labels = ['사망', '생존']

# Pie Plot
titanic['survived'].value_counts().plot.pie(ax=ax[0], explode=[0,0.1], shadow=True, autopct='%1.1f%%', labels=labels)
ax[0].set(ylabel='')
 
# Count Plot
sns.countplot(data=titanic, x='survived', ax=ax[1])
ax[1].set(xlabel='', xticklabels=labels)
 
plt.show()
```

```python
# 중복컬럼 제거
# sex 성별 - adult_male 성인남자여부 : 성별중복
# embarked - embark_town 탑승 항구 중복
# Survived - alive 생존여부
# pclass - class 좌석등급

titanic.drop(['sex','embark_town','alive','pclass'], axis=1, inplace=True)
```

```python
titanic.head()
```

```python
# 누락 데이터 개수 파악
titanic.isnull().sum()
```

```python
import missingno as msno

# 결측치 시각화
msno.matrix(titanic)
```

```python
# 결측치 처리
# embarked(2), age (177), deck(688)

titanic.embarked.value_counts()
```

```python
# embarked 변수는 결측치가 2개이므로 최빈값 S로 채우기.
titanic.embarked = titanic.embarked.fillna('S')
```

```python
# age 열에 데이터가 없는 모든 행을 삭제 (subset 옵션 이용)
titanic = titanic.dropna(subset=["age"])
```

```python
# deck 열의 NaN값을 바로 앞, 뒤의 값으로 변경 _ ffill
titanic['deck'].fillna(method='ffill', inplace=True)
titanic['deck'].fillna(method='backfill', inplace=True)  # 결측값 바로 아래 값과 동일하게 변경
```

```python
titanic.isnull().sum()
```

```python
# 파생변수
# 탑승한 가족 인원을 구한 family 컬럼 생성
titanic['family'] = 1 + titanic['sibsp'] + titanic['parch']
titanic.head()
```

```python
# 나이 범주화
def map_age(age):
  if age < 18 : 
    age = 0
  elif age < 25 : 
    age = 1
  elif age < 35 : 
    age = 2
  elif age < 45 : 
    age = 3
  elif age < 55 : 
    age = 4
  else : age = 5
  return age

titanic['age_new'] = titanic['age'].map(map_age)
titanic.head()
```

```python
# (age, sibsp, parch) age_new, family 생성 변수 삭제 drop
titanic.drop(['age','sibsp','parch','alone'], axis=1, inplace=True)
```

```python
titanic.head()
```

```python
#  탑승 클래스('class')별 생존율 비교
titanic[['class', 'survived']].groupby(['class'], as_index=False).mean().sort_values(by='survived', ascending=False)
```

-> 객실 등급이 좋을 수록 생존율이 높음

```python
#  성별('adult_male')별 생존율 비교
titanic[["adult_male", "survived"]].groupby(['adult_male'], as_index=False).mean().sort_values(by='survived', ascending=False)
```

-> 여성의 생존율이 남성보다 높음

```python
#  탑승 클래스('class')별 + 성별('Sex') 생존여부('Survived') 데이터 분포확인
sns.barplot(x='class', y='survived', hue='adult_male', data=titanic)
```

```python
#  family별 생존율 비교
titanic[["family", "survived"]].groupby(['family'], as_index=False).mean().sort_values(by='survived', ascending=False)
```

```python
# 훈련 자료에서 승선지(Embarked), 생존 여부(Survived), 성별(Sex)에 따른 요금(Fare)
grid = sns.FacetGrid(titanic, row='embarked', col='survived', height=2.2, aspect=1.6)

# 바그래프로 시각화, x: 성별, y: 요금, Error bar: 표시 안 함
grid.map(sns.barplot, 'adult_male', 'fare', alpha=.5, ci=None)
grid.add_legend()
```

-> 승선지가 C인 경우, 생존한 승객들의 평균 요금이 비교적 높음

```python
# class, who, embarked, adult_male, deck 범주화

# class
class_mapping = {"First": 1, "Second": 2, "Third": 3}
titanic['class'] = titanic['class'].map(class_mapping)
# adult_male
male_mapping = {True: 1, False: 0}
titanic['adult_male'] = titanic['adult_male'].map(male_mapping)
# who
who_map = {"man": 1, "woman": 2, "child": 3}
titanic['who'] = titanic['who'].map(who_map)
# enbarked
enbarked_map = {"S": 1, "C": 2, "Q": 3}
titanic['embarked'] = titanic['embarked'].map(enbarked_map)
```

```python
# deck 라벨인코딩
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
le = le.fit(titanic['deck'].unique())   #train['col']을 fit
titanic['deck'] = le.transform(titanic['deck'])   #train['col']에 따라 encoding
```

```python
# 클래스별 요금과 생존비교
sns.factorplot(x="class", y="fare", hue="survived", data=titanic[titanic.fare < 200], kind="box")
```

-> 가격이 높은 class 에 생존한 사람이 많다.

```python
# pd.cut을 이용한 fare_class 컬럼 생성
import pandas as pd

labels = ['1', '2', '3', '4', '5']
titanic['fare_class'] = pd.cut(titanic.fare, 5, labels=labels)
titanic['fare_class'].value_counts()
```
