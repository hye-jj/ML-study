```python
# 박스플롯 **by="species"**
iris.boxplot(by="species", figsize=(12, 6))
```

```python
# 산점도
sns.scatterplot(x="sepal_length", y="sepal_width", data=iris)
```

```python
# hue 인자 : 붖꽃 종류에 따라 데이터 포인트의 색을 다르게
sns.scatterplot(x='sepal_length', y='sepal_width', data=iris, hue='species')
```

```python
# 조인트플롯 : 산점도와 히스토그램을 동시에
sns.jointplot(x="sepal_length", y="sepal_width", data=iris, size=5)
```

```python
# FacetGrid : 서로 다른 스타일의 서브 플롯을 그릴 때
ax = sns.FacetGrid(iris, hue="species", size=5)\
    .map(plt.scatter, "sepal_length", "sepal_width")\
    .add_legend()
ax = sns.jointplot(x="sepal_length", y="sepal_width", data=iris, size=5)
```

```python
# 박스플롯
sns.boxplot(x="species", y="petal_length", data=iris)
```

```python
# 스트립플롯 : jitter=True로 설정해야 데이터 포인트가 퍼집니다
sns.stripplot(x="species", y="petal_length", data=iris, jitter=True, edgecolor="gray")
```

```python
# 바이올린플롯 
sns.violinplot(x="species", y="petal_length", data=iris, size=20)
```

```python
# 페어플롯 : 피처끼리의 상관관계 확인
sns.pairplot(iris, hue="species", size=3, diag_kind='hist') # diag_kind="kde"
```

```python
sns.FacetGrid(iris, hue="species", size=6) \
   .map(sns.kdeplot, "petal_length") \
   .add_legend()
```

```python
# 앤드류 곡선
from pandas.plotting import andrews_curves

plt.figure(figsize=(8,6))
andrews_curves(iris, "species")
```

```python
# 각 피처를 2차원 평면에 투영해서 데이터 샘플끼리의 상대적인 위치를 시각화
from pandas.plotting import radviz
plt.figure(figsize=(8,6))
radviz(iris, "species")
```
