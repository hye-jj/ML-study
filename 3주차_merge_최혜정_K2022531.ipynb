- 2주 수업을 진행한 줌 참여자료를 이용하여, 출석부에 표기를 합니다.

- 출석시 O를 결석시 X로 표기합니다. 표기된 결과를 "전자출석부 텍스트 출석부_기계학습기초_01분반.xls"에 저장합니다.

```python
# 출석부
df = pd.read_excel("/content/drive/MyDrive/대학원/202209자료/1)기계학습기초/data/전자출석부 텍스트 출석부_기계학습기초_01분반.xls")
```

```python
df.dropna(how = 'all', inplace=True)
df.head(2)
```

```python
# df 1주차 넣기
data = pd.merge(df,df1, how='outer',on=['이름','학번'], indicator=True)
data.loc[(data._merge == 'both'),('1 주차')] = 'O'
```

```python
data.drop(['_merge','역할'],axis=1, inplace=True)
data.head()
```

```python
data1 = pd.merge(data, df2, how='outer',on=['이름','학번'], indicator=True)
data1.loc[(data1._merge == 'both'),('2 주차')] = 'O'
data1.drop(['_merge','역할'],axis=1, inplace=True)
data1.dropna(inplace=True)
data1
```

```python
data1.to_excel('/content/drive/MyDrive/대학원/202209자료/1)기계학습기초/과제/전자출석부 텍스트 출석부_기계학습기초_01분반.xls', engine='openpyxl')
```
